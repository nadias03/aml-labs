```python
# kernel gaussian
def gaussian_kernel(x, x_i, h):
    u = (x - x_i) / h

    return (
        1 / np.sqrt(2*np.pi)
    ) * np.exp(-0.5*u**2)
```

```python
# kernel epanechnikov

def epanechnikov_kernel(x, x_i, h):
    u = (x - x_i) / h

    return np.where(
        np.abs(u) <= 1,
        0.75*(1-u**2),
        0
    )
```
